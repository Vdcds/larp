```python
# PR3: Informed Search Algorithms

# * Common: Graph representation
graph = {
    'A': [('B', 1), ('C', 4)],
    'B': [('D', 2), ('E', 5)],
    'C': [('F', 2)],
    'D': [('G', 3)],
    'E': [('G', 1)],
    'F': [('G', 3)],
    'G': []
}

# * Common: Heuristic values
heuristic = {
    'A': 6,
    'B': 5,
    'C': 4,
    'D': 3,
    'E': 1,
    'F': 2,
    'G': 0
}

start = 'A'
goal = 'G'


# * Common: Greedy Best First Search
def greedy_bfs(graph, heuristic, start, goal):
    open_list = [(heuristic[start], [start])]
    visited = set()

    while open_list:
        open_list.sort()
        h, path = open_list.pop(0)
        node = path[-1]

        if node == goal:
            return path

        if node not in visited:
            visited.add(node)

            for neighbour, cost in graph[node]:
                if neighbour not in visited:
                    open_list.append(
                        (heuristic[neighbour], path + [neighbour])
                    )

    return None


greedy_path = greedy_bfs(graph, heuristic, start, goal)
print("Greedy Best First Search:", " -> ".join(greedy_path))


```

```python
# * Common: A* Search
def astar(graph, heuristic, start, goal):
    open_list = [(heuristic[start], 0, [start])]
    visited = set()

    while open_list:
        open_list.sort()
        f, g, path = open_list.pop(0)
        node = path[-1]

        if node == goal:
            return path, g

        if node not in visited:
            visited.add(node)

            for neighbour, cost in graph[node]:
                if neighbour not in visited:
                    new_g = g + cost
                    new_f = new_g + heuristic[neighbour]
                    open_list.append(
                        (new_f, new_g, path + [neighbour])
                    )

    return None, 0
astar_path, total_cost = astar(graph, heuristic, start, goal)
print("A* Search:", " -> ".join(astar_path))
print("Total Path Cost:", total_cost)
```

```python
# * Common: Display results
greedy_path = greedy_bfs(graph, heuristic, start, goal)
astar_path, total_cost = astar(graph, heuristic, start, goal)

print("Greedy Best First Search:", " -> ".join(greedy_path))
print("A* Search:", " -> ".join(astar_path))
print("Total Path Cost:", total_cost)
```
