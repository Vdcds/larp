```python

# PR4: Knowledge Representation and Expert System

# * TASK 1: Define knowledge base
knowledge_base = {
    "flu": ["fever", "cough", "body_pain"],
    "cold": ["cough", "sneezing", "runny_nose"],
    "infection": ["fever", "body_pain"],
    "doctor_visit": ["high_fever", "breathing_problem"]
}


# * TASK 2: Take user input
print("MEDICAL EXPERT SYSTEM")
print("---------------------")

print("Enter your symptoms (yes/no):")

symptoms = [
    "fever",
    "cough",
    "body_pain",
    "sneezing",
    "runny_nose",
    "high_fever",
    "breathing_problem"
]

facts = []

for symptom in symptoms:
    answer = input("Do you have " + symptom.replace("_", " ") + "? ").lower()

    if answer == "yes":
        facts.append(symptom)


# * TASK 3: Apply Forward Chaining
print("\nInitial Facts:", facts)

conclusions = []

for disease, conditions in knowledge_base.items():

    if all(condition in facts for condition in conditions):
        conclusions.append(disease)


# * TASK 4: Display results
print("\nExpert System Results:")

if conclusions:
    for result in conclusions:
        print("Possible condition:", result.replace("_", " ").title())
else:
    print("No matching condition found.")

print("\nNote: This is only a simple educational demonstration.")
print("Please consult a doctor for medical advice.")


# * TASK 5: Completion
print("\nExpert System completed successfully!")
```
